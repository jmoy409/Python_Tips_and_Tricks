```python
class Robot:
    #if we run this function below on an object: john=, then will print out my name is john
    #self will refer to that object.
    #looks like a regular funcion, but when the function is within a class or a method of as class, need to add an additional argument: self
    #Need to add self to every argument for each function that you create
    def introduce_self(self):
        print("My Name is " + self.name) #self will refer to the object we are running this function on
```

```python
#can now create an object out of the class Robot
r1 = Robot()

#to set the attributes
r1.name = 'Tom'
r1.color = 'Red'
r1.weight = 30

#another robot object with different characteristics
r2= Robot()
r2.name = 'Jerry'
r2.color = 'Blue'
r2.weight = 50
```

```python
#now that we set the name attribute, we can use the introduce_self() function to print out the name
r1.introduce_self()
r2.introduce_self()
```

<h2>Looking back, we should have done things differently when adding attributes</h2>
<p>Instead of setting them up after defining the class, we should have done so when setting up the class</p>

```python
#can now create an object out of the class Robot
r1 = Robot()

### WHAT HAPPENS IF YOU ACCIDENTLY SET name attribute up as namee?
## we use name in introduce_self(), will depend on setting name up correctly
# Will create errors

#to set the attributes
r1.namee = 'Tom'
r1.color = 'Red'
r1.weight = 30

#another robot object with different characteristics
r2= Robot()
r2.name = 'Jerry'
r2.color = 'Blue'
r2.weight = 50
```

<h2>Solution: Constructor using __init__ in class definintion</h2>

```python
class Robot:
    #define constructor
    #need to add self keyword to every function and constructor
    #ensure argument name aside from self are defined in function
    def __init__(self, name, color, weight):
        #set that object's name to a given name
        self.name = name
        self.color = color
        self.weight = weight
    
    
    #if we run this function below on an object: john=, then will print out my name is john
    #self will refer to that object.
    #looks like a regular funcion, but when the function is within a class or a method of as class, need to add an additional argument: self
    #Need to add self to every argument for each function that you create
    def introduce_self(self):
        print("My Name is " + self.name) #self will refer to the object we are running this function on
```

```python
#much cleaner, dont need to specify attribute names every time
r1 = Robot("Tom", "red", 30)
r2 = Robot("Jerry", "blue", 40)
```

```python
r1.introduce_self()
r2.introduce_self()
```
