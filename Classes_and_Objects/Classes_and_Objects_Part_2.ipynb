```python
class Robot:
    #define constructor
    #need to add self keyword to every function and constructor
    #ensure argument name aside from self are defined in function
    def __init__(self, name, color, weight):
        #set that object's name to a given name
        self.name = name
        self.color = color
        self.weight = weight
    
    
    #if we run this function below on an object: john=, then will print out my name is john
    #self will refer to that object.
    #looks like a regular funcion, but when the function is within a class or a method of as class, need to add an additional argument: self
    #Need to add self to every argument for each function that you create
    def introduce_self(self):
        print("My Name is " + self.name) #self will refer to the object we are running this function on
```

```python
#Can create objects out of that class
r1 = Robot("Tom", "red", 30)
r2 = Robot("Jerry", "blue", 40)
```

```python
class Person:
    def __init__(self, n, p, i):
        self.name = n
        self.personality= p
        self.is_sitting = i
        #can do this or setting if after object has been created
        #p1.robot_owned=r2
        #make sure to make r as an argument in __init__
        #self.robotowned = r
        
    def sit_down(self):
        self.is_sitting=True
        
    def stand_up(self):
        self.is_sitting=False
        
```

```python
#can create a new object out of this class
p1 = Person("Alice", "aggressive", False)
p2 = Person("Becky", "talkative", True)
```

```python
#show that p1 owns r2
#can do this by setting robot owned attribute to p1
p1.robot_owned = r2
p2.robot_owned = r1
```
